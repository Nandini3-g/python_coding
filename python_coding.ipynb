# Problem: Find the First Non-Repeating Character
# You are given a string. Write a Python function to find the first character that appears only once in the string.

# If every character is repeated, return None.

#Examples
#Example 1:
#Input:  "programming"
#Output: 'p'


#Example 2:
#Input:  "aabbcdd"
#Output: 'c'


#Example 3:
#Input:  "aabbcc"
#Output: None


#Requirements
#Do not use external libraries.
#Maintain the original order of characters.
#Try to write an efficient solution.

#Bonus Challenges
#Solve it without using the count() method.
#Try solving it using a dictionary.
#Can you achieve O(n) time complexity? (Bonus)

```python
def non_repeating_character(s):
    d = {}
    for ch in s:
        if ch in d:
            d[ch]+= 1
        else:
            d[ch] = 1  
       
    
    for k in d.keys():
        
        if d[k] ==1 :
            return k
    print("None")
```

```python

non_repeating_character("programming")
```

```python
non_repeating_character("aabbcdd")
```

```python
non_repeating_character("aabbcc")
```

```python
l = [1,2,2,3,3,3,4,4]
def second_max(l):
    d = {}
    for i in l:
        d[i] = d.get(i,0)+1

    # finding second minimum
    fmax = float('-inf') 
    smax = float('-inf')

    for k,v in d.items():
        if v>fmax:
            smax = fmax
            fmax = v

        elif smax < v < fmax:
            smax = v

        
    for k,v in d.items():
        if v == smax:
    
            return k
    print(None)
```
